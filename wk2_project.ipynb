**Build a Guess the Number (with Difficulty Levels) Game**
Create a guessing game where the computer generates a random number and the user tries to guess it.
**Instructions:**
Use random.randint() to generate a number between 1 and 100.
Ask the user to input guesses and give feedback: “Too High”, “Too Low”.
Limit the number of guesses to 5–10 depending on the level.
**Features to implement:**
Allow user to choose difficulty: Easy (10 tries), Medium (7), Hard (5).
Count attempts and print remaining guesses.
End game on correct guess or when attempts run out.

```python
import random

def play_game():
    print("Welcome to the Number Guessing Game!")

    # choosing levels of difficulty
    while True:
        difficulty = input("Choose a difficulty. Type 'easy' (10 tries), 'medium' (7 tries), or 'hard' (5 tries): ").strip().lower()        
        if difficulty == "easy":
            max_attempts = 10
            break
        elif difficulty == "medium":
            max_attempts = 7
            break
        elif difficulty == "hard":
            max_attempts = 5
            break
        else:
            print("Invalid choice. Please type 'easy', 'medium', or 'hard'.")

    # initializing game variables
    secret_number = random.randint(1, 100)
    attempts_used = 0

    # game loop
    while attempts_used < max_attempts:
        print(f"Remaining guesses: {max_attempts - attempts_used}")

        try:
            guess = int(input("Enter your guess: "))
        except ValueError:
            print("Invalid input. Please enter a whole number.")
            continue

        if guess < 1 or guess > 100:
            print("Out of bounds! Guess a number between 1 and 100.")
            continue

        attempts_used += 1
        
    # check guess
        if guess == secret_number:
            print(f"Correct! The answer was {secret_number}. "
                f"You guessed it in {attempts_used} tries.")
            return
        elif guess > secret_number:
            print("Too High.")
        else:
            print("Too Low.")

    print(f"Game over! You've run out of guesses. The number was {secret_number}.")

if __name__ == "__main__":
    play_game()
           
```
