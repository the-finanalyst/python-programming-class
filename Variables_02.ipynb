**Where does data come from**    
We have worked with data stored in files such as: CSV files, excel spreadsheets, databases etc.
**Web Services**
- is a piece of software that makes itself available over the internet and uses a standardized messaging system.
- Allow different applications and systems to communicate & exchange info over the internet.
*Common ways to collect info from websites:
1. Accessing APIs (Application Programming Interface)
2. Web scraping
*Using an API*
A service may provide an `endpoint` that allows you to request information.
*Web scraping*
You could visit a website displaying the info, inspect its HTML structure & write a python script that extracts the info.
*How websites communicate*
*Where does information come from?
- Your browser sends a request to the server
- The server receives the request
- The server processes the request & retrieves the info.
- The server sends a response back to the browser
- The browser displays the returned content

-The communication follows a set of rules called **HTTP(Hypertext Transfer Protocol)**
HTTP defines how clients & servers communicate over the web.
*Components to understand*
**Client** The application requesting info, e.g your browser or python script.
**Server** The system or computer that receives the request, processes it & sends back a response.
**Database** An organised place where data collected is stored electronically

**URL(Uniform Resource Locator)** Is the address used to locate a resource on the internet.
**API endpoint** A specific URL through which an API makes a particular resource available.

**HTTP methods**
- They describe the action a client wants to perform when communicating with a server.
GET(Retrieves info)
POST(Send info to that resource)
PUT(Replace existing resourse)
DELETE(Remove a resource)

*Status codes*
After sending a request, the server returns a response containing a status code.
- 200: Ok, the request sent was successful.
- 201: Your resource was created successfully
- 400: Bad request - not valid

````python
class BankAccount:
    def __init__(self, account_holder, balance):
        self.account_holder = account_holder
        self.balance = balance

    def deposit(self, amount):
        self.balance += amount
        print(f"Deposited {amount}")

    def withdraw(self, amount):
        if amount <= self.balance:
            self.balance -= amount
            print(f"withdrawn {amount}")
        else:
            print("You are broke")

    def display_balance(self):
        print(f"Balance: {self.balance}")


account1 = BankAccount("Mark", 500)
account1.deposit(300)
account1.display_balance()
account1.withdraw(100)
````

**Instance Attributes**
These are variables that are unique to each specific object.

-Usually defined inside the __init__() using self
-Belong to individual objects

**Class Attributes**
are shared among all instances of a class.

-Defined directly inside the class
-Shared at the class level


*Pillars of OOP
**Encapsulation**
 is the practice of bundling Data (Attributes) and Methods (Behaviors) into a single unit (a Class) and restricting direct access to some of that object's components.
 Controls how attributes are accessed and modified.


**Abstraction**
 is the principle of hiding the complex implementation details and showing only the essential features of an object.
It allows the user to interact with a system using a clean, simple Interface without needing to understand the complicated logic happening in the background.


**Inheritance**
is a fundamental mechanism in Object-Oriented Programming that allows a new class to acquire the properties (attributes) and behaviors (methods) of an existing class.

-Promotes code reusability

**Polymorphism** 
Allows different objects to be treated as the same type of thing, even if they are completely different classes. It allows us to perform the same action in different ways depending on the object.

````python
#Inheritance
class Employee:
    def __init__(self, name, salary):
        self.name = name
        self.salary = salary
    def display_details(self):
        print(self.name, self.salary)

class DataAnalyst(Employee):
    def analyst_data(self):
        print(f"{self.name} is analyzing data")
class SoftwareDeveloper(Employee):
    def write_code(self):
        print(f"{self.name} is writing code")

analyst = DataAnalyst("Mark", 100000)
developer = SoftwareDeveloper("Mike", 150000)

analyst.display_details()     
developer.display_details()   

developer.write_code()
analyst.analyst_data()


````

````python
#Encapsulation
class BankAccount:
    def __init__(self, account_holder, balance, pin):
        self.account_holder = account_holder #public
        self._balance = balance #protected
        self.__pin = pin #private

    def dislay_account(self):
        print(f"Account Holder: {self.account_holder}")
        print(f"Balance: {self._balance}")

    def verify_pin(self, pin):
        return self._pin == pin

    def view_pin(self):
        print(self.__pin)


    def deposit(self, amount):
        self._balance += amount
        print(f"Deposited: {amount}. Current Balance: {self._balance}" )

    def withdraw(self, amount):
         if amount <= self._balance:
            self._balance -= amount
            print(f"Withdrawn: {amount}")
         else:
             print("Insufficient balance")
        
        
      
account = BankAccount("Melody", 5000, 1234)

#Accessing public attribute
account.account_holder

#Accessing protected attribute
account.dislay_account()
account._balance

#Accessing the private attribute
account.view_pin()

#Updating
account.account_holder = "Mike"
account

account.deposit(2000)
account.dislay_account()

account.withdraw(1000)
account.dislay_account()

````

````python
#Abstraction
#Payment System using Abstract classes
from abc import ABC, abstractmethod

class Payment(ABC):
    @abstractmethod
    def pay(self, amount):
        pass

class MpesaPayment(Payment):
    def pay(self, amount):
       print(f"Processing Mpesa payment of Ksh {amount}")
       print(f"Payment Successful")

class CardPayment(Payment):
    def pay(self, amount):
        print(f"Processing Card payment of Ksh {amount}")
        print(f"Payment Successful")

payment1 = MpesaPayment()
Payment2 = CardPayment()

payment1.pay(6500)
````
