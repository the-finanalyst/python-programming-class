#Understanding Regex Patterns
Is made up of characters & symbols that describe what we want to find.
1. Literal characters match themselves 
2. Special Regex Symbols have meanings
**Literal characters**
Ordinary characters that match exactly what they reperesent.
**Metacharacters**
Symbols that have a special meaning in regex. Allows us search for patterns instead of exact characters.


**(.)** Matches any single character

**Character class []**
Allow us to specify which characters should be matched.

**Negated Character Class ^...** 
Matches characters not listed inside the brackets
[^abc]d, e, f
**()** Groups part of a pattern together
**\\|**  Matches either pattern (OR)
**|\|**  Escapes a special character or introduces a special sequence.


**$** End of a string


````python
tex = "My student ID is 2028761"
pattern = re.search(r"\d+", text)
result = pattern.group()
print(result)
````

````python
text = "I am learning Python"
result = re.search(r"Python", text)
print(result.group())
````

````python
# re.search() 1st occurrence of a pattern in the txt, If no match returns None
message = "Payment reference: TXN-84721. Status: complete"
result = re.search(r"TXN-\d+", message)
print(result.group())
````

````python
#re.match() checks a pattern at the begginning of a string.
reference = "INV-2048 REF-2078 INV-456"
result = re.match(r"INV", reference)
print(result.group())
````

````python
# re.fullmatch()	
# Checks whether the entire text matches the pattern
product_code = "AB2048"
result = re.fullmatch(r"[A-Z]{2}\d{4}", product_code)
print(result.group())
````

````python
#re.finditer()	
#Finds all matches and returns them one at a time
#An iterator of Match Objects
message = "Error 404 occurred. Error 503 occurred."
result = re.finditer(r"\d{3}", message)
for match in result:
    print(match.group())
````

````python
# re.sub()- Replaces matching text (returns a string)
# Pattern- Replacement- text
message = "call 0715869454 or 0712457900 for assistance"
result = re.sub(r"\d+", "0715869454", message)
print(result)
````

````python
text = "Product: Laptop!!1 Category:@Electronics"
result = re.sub(r"[!@]", " ", text)
print(text)
````

````python
#re.split() Splits text wherever the pattern matches
# returns a text
tags = "Python, SQL: Power BI | Excel"
result = re.split(r"[,;|]\s*", tags)
print(result)
````

````python
# re.compile() Creates a reusable Regex pattern
# Instead of writing the same pattern repeatedly, we write it once repeatedly
message_1 = "Order ORD-1024 has been shipped"
message_2 = "Order ORD-1024 has been packed"

pattern = re.compile(r"ORD-\d{4}")

print(pattern.findall(message_1))
````

** Grouping & Extacting Information**

````python
text = "PEN-104:25, BOOK-208:12, BAG-310:8"
result = re.findall(r"([A-Z]+-\d+):(\d+)", text)
print(result)
````

````python
# Use appropriate regex functions to extract all ticket IDs
# Find 1st ticket ID
# Replace ticket IDs with hidden
# Split the msg at each comma
message = "Ticket TK-204 is open, while ticket TK-518 is closed"
tickets = re.findall(r"TK-\d+", message)
print(tickets)

first_ticket = re.search(r"TK-\d+", message)
print(first_ticket.group())

hidden_message = re.sub(r"TK-\d+", "HIDDEN", message)
print(hidden_message)  

split_message = re.split(r",\s*", message)
print(split_message)
````

**#Advanced Regex Concepts**
1. Lookahead - checks whether a pattern is followed by another pattern without including the 2nd pattern in the returned match.
2. Lookbehind - Checks whether a pattern is preceeded by another pattern
3. Greedy Matching

````python
#Lookahead
text = "Prices: 500 KES, 1000 USD, 250 KES"
result = re.findall(r"\d+(?= KES)", text)
print(result)
````

````python
#Lookbehind
#Extract numbers that appear after KES
text = "Prices: KES 500, USD 1000, KES 250 "
result = re.findall(r"(?<=KES )\d+", text)
print(result)
````

````python
#Greedy Matching
text = "Items: [Laptop][Mouse][Keyboard]."
result = re.findall(r"\[.*]", text)
print(result)
````
