A business offers a discount to customers who are either :

- Registered students
- Senior citizens aged 60 and above

Write a program that determines whether a customer qualifies

```python
registered_students = input("are you a registered student? (yes/no): ")
senior_citizens_age = int(input("Enter your age: "))
if registered_students == "yes" and senior_citizens_age >= 60:
    print("You are eligible for a discount.")
else:
    print("You are not eligible for a discount.")
```
