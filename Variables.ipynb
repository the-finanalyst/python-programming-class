## Lambda Function
Its a small anonymus function defined using the `lambda` keyword
Written in a single expression & do not require a function name in their definition.
**Syntax**
```python
lambda arguments: expression
e.g Squaring a number

````python
# Using a Regular Function
def square(number):
    return(number(math.pow(2)))
result = square(4)
print("Result:", result)
````

````python
square = lambda number: number **2
print(square(4))
````

````python
add = lambda a, b: a + b 
add(10, 20)
````

Lambda with map()

````python
numbers = [1, 2, 3, 4, 5]
square = list(map(lambda digit: digit **2, numbers))
print(square)
````

Lambda with Filter()

````python
numbers = [10, 15, 20, 25, 30, 35]
even_numbers = list(filter(lambda x: x % 2 == 0, numbers))
print(even_numbers)
````
