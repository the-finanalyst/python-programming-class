## Exe 1.1
a.Create a program that acts as a simple calculator. It should take two numbers and an operator (+, -, *, /) as input and perform the corresponding arithmetic operation. Add comments to explain each part of the code.
b.Write a program that generates and prints a random number between 1 and 100. Use the random library.

```python
operator = input("Enter an operator (+, -, *, /): ")
number1 = float(input("Enter 1st number: "))
number2 = float(input("Enter 2nd number: "))
if operator == "+":
    result = number1 + number2
elif operator == "-":
    result = number1 - number2
elif operator == "*":
    result = number1 * number2
elif operator == "/":
    result = number1 / number2
else:
    result = "Invalid operator"     
print(f"Result: {result}")
```
