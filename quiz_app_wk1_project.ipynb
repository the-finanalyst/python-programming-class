## Command-Line Based Quiz Application
Build an interactive quiz app that runs in the terminal and scores users based on correct answers.
Instructions:
Create a list of questions. Each question should have:
The question text
Multiple choices (A, B, C, D)
The correct answer
Display each question one by one using input() to collect answers.
Use if statements to compare user answers to correct ones and assign scores.
Use loops to iterate through the questions.
At the end, display the total score and provide feedback.
Features to implement:
Use of random.shuffle() to mix question order.
Keep track of score and give feedback: (“Excellent”, “Good”, “Try Again” based on score).
Use try/except to handle invalid input.

```python
import random
def run_finance_quiz():
# List of Questions
    questions = [
        {"question": ("Which financial statement reports a company's revenues and expenses over a specific period?"),
          "choices": [
              "A. Balance Sheet",
              "B. Income Statement",
              "C. Cashflow Statement",
              "D. Statement of changes in Equity"],
          "answer": "B"},
        {"question": ("In Kenya, Companies reporting under the Companies Act are required to prepare their financial statements in compliance with:"),
                    "choices": [
                        "A. IFRS through ICPAK",
                        "B. US GAAP",
                        "C. Central Government through KRA",
                        "D. CBK"],
                    "answer": "A"},
        {"question": ("In Kenya, when is the national budget presented in the National Assembly?"),
                      "choices": [
                          "A. January",
                          "B. October",
                          "C. July",
                          "D. June"],
                      "answer": "D"},
        {"question": ("If a company purchases equipment using cash, what is the immediate effect on total assets?"),
                  "choices": [
                      "A. Decrease",
                      "B. Increase",
                      "C. Remain Unchanged",
                      "D. Decreased Liabilities"],
                     "answer": "C"}]
     #Shuffle the questions
    random.shuffle(questions)
    
    score = 0
    total_questions = len(questions)
    valid_choices = ("A", "B", "C", "D")
    
    print("FINANCIAL ANALYSIS QUIZ APP")
    print("Instructions: Type A, B, C or D and press Enter.\n")
    
     # Iterating through the questions
    for index, item in enumerate(questions, start=1):
        print(f"Question {index}/{total_questions}: {item['question']}")
        for choice in item["choices"]:
          print(f"   {choice}")

    # Using try/except functions
        while True:
            try:
             user_answer = input("Your answer (A/B/C/D):").strip().upper()
             if user_answer not in valid_choices:
                raise ValueError(f"Invalid Selection: Please enter A, B, C, or D.")
             break
            except ValueError as err:
             print(err)   

    # Updating Scores
        if user_answer == item["answer"]:
         print("Correct!\n")
         score += 1
        else:
         print(f"Incorrect. The correct answer is {item['answer']}.\n")  

    # Providing Feedback
    percentage = (score / total_questions) * 100
    if percentage >= 80:
        feedback = "Excellent"
    elif percentage >= 50:
        feedback = "Good"
    else:
         feedback = "Try Again"

    # Final Results
    print(f"    QUIZ FINAL RESULTS")
    print(f"Total Score : {score} out of {total_questions} ({percentage:.2f}%)")
    print(f"Feedback    : {feedback}")

if __name__ == "__main__":
        run_finance_quiz()
```
