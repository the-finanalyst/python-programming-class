Working with SQLite in Python
SQLite- A lightweight relational database that stores data in python
Python includes a built-in-module called sqlite3 for working with SQLite databases

```python
# Connecting to a database
import sqlite3
```

```python
connection = sqlite3.connect("school.db")
#create a cursor(allows us to excecute SQL commands)
cursor = connection.cursor()
```

```python
cursor.execute("""
    CREATE TABLE IF NOT EXISTS
    students(
        id INTEGER PRIMARY KEY AUTOINCREMENT,
        name TEXT,
        age INTEGER,
        course TEXT)

""")
connection.commit()
```

```python
#Inserting records
cursor.execute(
    "INSERT INTO students(name, age, course)VALUES (?, ?, ?)", 
    ("Jane", 25, "Data Science")
)
connection.commit()
```

```python
#Inserting multiple records
students_data = [("Peter", 19, "Data Science"),
                 ("Mary", 27, "Python"),
                 ("Tom", 30, "Data Analytics")
]
cursor.executemany(
    "INSERT INTO students(name, age, course) VALUES (?,?,?)",
    students_data
)
connection.commit()
```

```python
#Retrieve records
cursor.execute("SELECT * FROM students")
students = cursor.fetchall()
for student in students:
    print(student)
```

```python
#Filter records
cursor.execute(
    "SELECT * FROM students WHERE age > ?",
    (18,)
)
print(cursor.fetchall())
```

```python
# Update records
cursor.execute(
    "UPDATE students SET age = ? WHERE name = ?",
    (27, "Peter")
)
connection.commit()
```

```python
#Delete
cursor.execute(
    "DELETE FROM students WHERE name = ?",
    ("Peter")
)
connection.commit()
```

**Creating related tables in SQLite**

```python
import sqlite3
```

```python
#Connection to the database
connection = sqlite3.connect("online_store.db")
cursor = connection.cursor()
```

```python
# Enable foreign key constraints
cursor.execute("PRAGMA foreign_keys =ON")
```

```python
cursor.execute(""" 
    DROP TABLE customers
""")
connection.commit()
```

```python
#create the customers table
cursor.execute(""" 
CREATE TABLE IF NOT EXISTS customers (
    customer_id INTEGER PRIMARY KEY AUTOINCREMENT,
    customer_name TEXT,
    email TEXT
    )
""")
connection.commit()
```

```python
#create the order table
cursor.execute("""
    CREATE TABLE IF NOT EXISTS orders (
    order_id INTEGER PRIMARY KEY AUTOINCREMENT,
    product TEXT,
    amount FLOAT,
    customer_id INTEGER,
    FOREIGN KEY(customer_id) REFERENCES customer (customer_id)
    )
""")
connection.commit()
```

```python
#Insert customers
customers_data = [
    ("John", "john@gmail.com"),
    ("Alice", "alice@gmail.com"),
    ("Peter", "peter@gmail.com"),
    ]
cursor.executemany(
    "INSERT INTO customers (customer_name, email) VALUES (?, ?)", 
    customers_data
)
connection.commit()
```

```python
#Insert into the orders table
orders_data = [
    ("Laptop", 12000.0, 1),
    ("Charger", 500.0, 2),
    ("Headphones", 1000.0, 3),
]
cursor.executemany(
  "INSERT INTO orders (product, amount, customer_id) VALUES (?, ?, ?)", orders_data  
)
connection.commit()
```

```python
cursor.execute("""
INSERT INTO orders (product, amount, customer_id) VALUES (?, ?, ?)
""", ("CD", 400, 2))
connection.commit
```

```python
#Combining related tables using joins
cursor.execute("""
    SELECT
        customers.customer_name,
        orders.product,
        orders.amount
    FROM customers
    JOIN orders
    ON customers.customer_id = orders.customer_id
""")
for record in cursor.fetchall():
    print(record)
```
