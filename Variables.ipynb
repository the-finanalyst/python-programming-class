## Loop Patterns
# Counting
# summation
# Maximum value


````python
numbers = [1, 2, 3, 4, 5]
count = 0

for numb in numbers:
    if numb < 5:
        count +=1

    print("Count:", count)
````

````python
numbers = [1, 2, 3, 4, 5]
total = 0
for numb in numbers:
    total += numb
print("Total:", total)
````

````python
numbers = [1, 2, 3, 4, 5]
maximum = numbers[0]
for num in numbers:
    if num > maximum:
        maximum = num
print("Maximum:", maximum)
````
